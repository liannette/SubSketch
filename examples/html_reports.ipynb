```python
from pathlib import Path
from IPython.display import HTML, display

import subsketch as subsk
```

First generate a session to load all data. This ensures that all input files need be read just once.

```python
session = subsk.SubSketchSession(
    motifs_file="example_data/clusterclue_motifs.txt",
    genbank_dir="example_data/mibig_4.0_gbks",
    domain_hits_file="example_data/mibig_4.0_domain_hits.txt",
    motif_hits_file="example_data/mibig_4.0_detected_motifs.tsv",
    compounds_file="example_data/mibig_4.0_compounds.tsv",
)
session.load()
```

# HTML report for a BGC

```python
# Full report with all plots

bgc_id = "BGC0000893"
html_content = session.html_report_for_bgc(
    bgc_id=bgc_id,
    gene_arrow_scaling=60
)

display(HTML(html_content))

# save to file
output_filepath = Path("example_results") / f"{bgc_id}_report_full.html"
with open(output_filepath, "w") as f:
    f.write(html_content)
```

```python
# Minimal report witout optional plots

bgc_id = "BGC0000893"
html_content = session.html_report_for_bgc(
    bgc_id=bgc_id,
    include_bgc_plot=False,
    include_compound_plots=False,
    include_motif_plots=False,
    gene_arrow_scaling=60
)

display(HTML(html_content))

# save to file
output_filepath = Path("example_results") / f"{bgc_id}_report_minimal.html"
with open(output_filepath, "w") as f:
    f.write(html_content)
```

```python
# Full report per motif

motif_id = "M1181"
html_content = session.html_report_for_motif(
    motif_id=motif_id,
    gene_arrow_scaling=60
)

display(HTML(html_content))

# save to file
output_filepath = Path("example_results") / f"Motif{motif_id}_report_full.html"
with open(output_filepath, "w") as f:
    f.write(html_content)
```
